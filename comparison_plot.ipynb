```python
import os
import plotly
import plotly.graph_objects as go
import pandas as pd
import plotly.colors as pc
```

```python
comparison_plot_dir = './comparison_plots'
os.makedirs(comparison_plot_dir, exist_ok=True)

topic_dir = os.path.join(comparison_plot_dir, 'original arches')
os.makedirs(topic_dir, exist_ok=True)

topic_plot_dir = os.path.join(topic_dir, 'comparison_plot')
os.makedirs(topic_plot_dir, exist_ok=True)

pdf_save_dir = os.path.join(topic_plot_dir, 'pdf')
os.makedirs(pdf_save_dir, exist_ok=True)
html_save_dir = os.path.join(topic_plot_dir, 'html')
os.makedirs(html_save_dir, exist_ok=True)

# Then put models in this directory
models_dir = os.path.join(topic_dir, 'models')
os.makedirs(topic_dir, exist_ok=True)
```

```python
category_dict = {'se':'spherical_error',
                 'ae':'azimuth_error',
                 'ee':'elevation_error',
                 'ca':'combined_accuracy',
                 'aa':'azimuth_accuracy',
                 'ea':'elevation_accuracy'}

for category in category_dict.values():
    # if category == category_dict['se']:
        models_list = sorted([m for m in os.listdir(models_dir) if m != '.DS_Store'])
        print(f'models_list : {models_list}')
        n = len(models_list)
        colors = pc.sample_colorscale('Cividis', [i / (n-1) for i in range(n)])
        fig = go.Figure()
        for i, model in enumerate(models_list):
            model_dir = os.path.join(models_dir, model)
            # Create subplots
            # Plot 1: Angular Errors by SNR
            results_by_snr = pd.read_csv(os.path.join(model_dir, 'results_by_snr.csv'))
            print(results_by_snr)

            snr_values = results_by_snr['snr']
            metric_values = results_by_snr[category]

            if category == 'spherical_error':
                color = '#D62728'
            elif category == 'azimuth_error':
                color = '#1f77b4'
            elif category == 'elevation_error':
                color = '#ff7f0e'

            fig.add_trace(go.Scatter(
                x=snr_values, y=metric_values,
                mode='lines+markers',
                name=f'{model}',
                line=dict(color=colors[i], width=2),
                marker=dict(size=8)
            ))

        yaxis_title = f'{category} (degrees)' if 'error' in category else f'{category}'
        fig.update_layout(
            title=dict(text=f"{category} by SNR", font=dict(size=20)),
            xaxis=dict(title="SNR (dB)", tickfont=dict(size=12), showgrid=True, gridcolor='lightgray', zeroline=False),
            yaxis=dict(title=yaxis_title, tickfont=dict(size=12), showgrid=True, gridcolor='lightgray'),
            width=1000,
            height=600,
            paper_bgcolor="white",
            plot_bgcolor="white",
            legend=dict(x=0.7, y=0.95)
        )
        # Save pdf / html separately
        fig.write_image(os.path.join(pdf_save_dir, f"{category}_by_snr.pdf"))
        fig.write_html(os.path.join(html_save_dir, f"{category}_by_snr.html"))
```
